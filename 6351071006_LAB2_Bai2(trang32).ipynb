**Đề bài: Phát hiện bệnh sốt xuất huyết**

Một mô hình AI được dùng để phát hiện bệnh **sốt xuất huyết** trên 120 bệnh nhân. Kết quả thu được như sau:

- **Phát hiện đúng 45 bệnh nhân bị sốt xuất huyết** (True Positive – TP)  
- **Phát hiện nhầm 8 bệnh nhân khỏe thành bị bệnh** (False Positive – FP)  
- **Bỏ sót 12 bệnh nhân bị sốt xuất huyết** (False Negative – FN)

**Yêu cầu:** Tính **F-measure** (F1-score) của mô hình.


```python
print(STUDENT)

from sklearn.metrics import precision_score, recall_score, f1_score

# Tổng số bệnh nhân bị sốt xuất huyết: TP + FN = 45 + 12 = 57
# Tổng số bệnh nhân khỏe: FP + TN = 8 + TN → tổng 120 → TN = 120 - 57 - 8 = 55

y_true = [1] * 57 + [0] * 63
y_pred = [1] * 53 + [0] * 12 + [0] * 55

precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print(f"Giá trị Precision: {precision:.3f}")
print(f"Giá trị Recall: {recall:.3f}")
print(f"Giá trị F1-score (F-measure): {f1:.3f}")
```
